# Structured Output in AI Engineering

This notebook explains what structured output means for AI systems and shows three common Python approaches: Pydantic, TypedDict, and dataclasses.

## What is structured output?

Structured output means the model returns a predictable, machine-readable format instead of an unstructured text response. It is useful for downstream code, data validation, and safe parsing.

In AI engineering, structured output helps you: 
- enforce schema expectations,
- validate fields and types,
- convert model responses into objects, and
- integrate with APIs or databases more reliably.

## Different ways to represent structured output in Python

We will look at three patterns: Pydantic models, TypedDicts, and dataclasses. Each has a different level of validation, typing support, and runtime behavior.

## 1. Pydantic

### What is it?
Pydantic is a Python library for data parsing and validation using Python type hints. It creates classes that enforce types at runtime and convert inputs automatically.

### Input validation
Pydantic checks incoming values and raises errors for invalid fields. That makes it ideal when you need strong guarantees about AI outputs.

### Nested structures
Pydantic supports nested models, lists, and dictionaries, so you can represent rich structured data cleanly.

### Description
Pydantic models can include field descriptions and metadata. These descriptions are useful for documentation and can also be used by prompt engineering tools to describe expected output.

```python
from pydantic import BaseModel, Field, ValidationError
from typing import List

class Address(BaseModel):
    street: str = Field(..., description="Street address")
    city: str
    zip_code: str

class PersonOutput(BaseModel):
    name: str
    age: int
    hobbies: List[str]
    address: Address

# Example structured output from an AI model
raw = {
    "name": "Aisha",
    "age": "30",  # string input will be coerced to int
    "hobbies": ["reading", "hiking"],
    "address": {
        "street": "123 Main St",
        "city": "Seattle",
        "zip_code": "98101"
    }
}

try:
    person = PersonOutput(**raw)
    print(person.json(indent=2))
except ValidationError as exc:
    print("Validation failed:")
    print(exc.json(indent=2))
```

## 2. TypedDict

### What is it?
TypedDict is a typing construct from the `typing` module that describes the expected structure of a dictionary. It is used only for static type checking and does not validate data at runtime.

### How is it different?
TypedDict is lighter weight than Pydantic and works well when you want IDE support and type hints without runtime enforcement. It does not coerce values or raise validation errors by itself.

```python
from typing import TypedDict, List

class PersonDict(TypedDict):
    name: str
    age: int
    hobbies: List[str]

# Example structured output shape
person: PersonDict = {
    "name": "Carlos",
    "age": 28,
    "hobbies": ["cycling", "travel"]
}

print(person)
```

## 3. Dataclasses

### What is it?
`dataclasses` is a standard library module that creates simple immutable or mutable classes for structured data. It is useful when you want plain Python objects with automatic `__init__`, `__repr__`, and comparison methods.

### How is it different?
Dataclasses are runtime objects and can be instantiated directly. They do not provide automatic validation by default, but they are easy to use and expressive for structured results.

```python
from dataclasses import dataclass
from typing import List

@dataclass
class PersonData:
    name: str
    age: int
    hobbies: List[str]

# Example structured output as a dataclass
person = PersonData(name="Priya", age=34, hobbies=["painting", "gardening"])
print(person)
print(person.name)
```
