# Messages in LangChain and AI Engineering

This notebook explains what messages are in the context of AI chat systems, why they matter, and the main types you commonly use in LangChain.

## What are messages?

A message is the basic unit of communication between the user, the system, and the AI model. In chat-based AI systems, a conversation is represented as a sequence of messages. Each message includes a role and content, and optionally metadata or additional properties.

Messages are important because they define the prompt structure, guide the model's behavior, and preserve the conversation history needed for multi-turn interactions.

## Why messages matter in LangChain

1. **Structured prompts:** LangChain sends messages to language models in a structured way, so role and content are clearly separated.
2. **Prompt engineering:** Different message roles help you control tone, instructions, and user questions.
3. **Memory and chains:** Message sequences can be stored as memory or reused inside chains for follow-up queries.
4. **Tool use:** Tools and agents often rely on messages to decide what action to take next.

## Common message roles

- **system**: Provides the model with high-level instructions or context. This is where you set behavior like "You are a helpful assistant."
- **user**: Contains the user’s question or request. It represents what the end user is asking the model to do.
- **assistant**: Contains the model’s response or previous replies. It helps preserve conversation flow and context.
- **tool**: In LangChain and agent workflows, a message from a tool can represent tool output or observations.
- **function** / **tool_response**: Some systems use a separate role for structured function/tool outputs.

## Types of messages in AI engineering

### 1. Instruction messages
These are usually system messages that tell the model how to behave. Examples:
- `You are a helpful assistant.`
- `Answer as a polite technical tutor.`

### 2. User messages
These are queries or commands from the user. Examples:
- `Summarize the following text.`
- `Write a Python function that reverses a string.`

### 3. Assistant messages
These are responses generated by the model in previous turns. They preserve conversation context for multi-turn dialogs.

### 4. Tool messages
In agent workflows, tools can send messages back to the chain with structured data or results. Examples include search results, calculator output, or database query responses.

### 5. System / metadata messages
These include hidden context or instructions not visible to the user, such as API keys, locale settings, or special formatting rules. They help the model follow policy or domain constraints.

## Example message format

A simple chat sequence might look like this in JSON-style form:

```json
[
  { "role": "system", "content": "You are a helpful assistant." },
  { "role": "user", "content": "What is the capital of France?" },
  { "role": "assistant", "content": "The capital of France is Paris." }
]
```

In LangChain, these messages are used to build prompts, manage memory, and run chains or agents.

## Best practices

- Keep system messages clear and concise.
- Use user messages for real questions and tasks.
- Preserve assistant messages only when you need a multi-turn conversation.
- Separate tool output from user intent to avoid confusing the model.
- Avoid putting secrets or private data directly into prompts or message content.
